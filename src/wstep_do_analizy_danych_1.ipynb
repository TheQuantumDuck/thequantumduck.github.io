### Interludium na stringi

Wypada tutaj wyjaśnić szerzej, co dokładnie stało się w linijce 'path_to_data = path + "/who-suicide-statistics.csv"'. W języku Python tekst pisany w angielskich cudzysłowach jest typem danych odpowiadający zwykłemu tekstowi – 'string' (ang. *łańcuch*). Napis bez cudzysłowów – 'path', nie jest sam w sobie stringiem, a odniesieniem do pewnej wartości zapisanej wcześniej. To mamy na myśli, kiedy mówimy, że path jest zmienną. Kiedy używamy operatora dodawania na dwóch zmiennych typu 'string', to wynikiem jest 'string' powstały ze zlepienia ich ze sobą. Połączenie ścieżki zapisanej w zmiennej 'path' i nazwy pliku poprzedzonej ukośnikiem zapisaliśmy do nowej zmiennej – 'path_to_data'. Co do zasady nazwy zmiennych powinny zawierać jedynie litery angielskiego alfabetu i podkreślniki. Używanie innych znaków jest albo zakazane, albo niepraktyczne.

## Do roboty

Teraz jesteśmy gotowi zabrać się za dane. Do ich wczytywania i transformacji będziemy używać biblioteki 'Polars'. Ponieważ plik, który mamy, jest plikiem 'csv', to użyjemy funkcji stosownej do tego rodzaju pliku. W drugiej linijce użyjemy funkcji `head`, której zadaniem jest wyświetlenie pierwszych kilku wierszy danych.

```python
df = pl.read_csv(path_to_data)
df.head(5)
```

Udało się! Wczytaliśmy dane i zapisaliśmy do zmiennej 'df'. Ta zwyczajowa nazwa bierze się stąd, że typ danych odpowiadający takiej tabeli, to 'DataFrame'. Widzimy powyżej wszystkie sześć kolumn, jakie znajdują się w danych. Pod ich nazwami są skróty sygnalizujące jaki typ danych znajduje się w każdej z nich. Jak łatwo się domyślić 'str' odpowiada stringom. Druga sygnatura – 'i64', wskazuje na wartości typu 'integer', czyli liczby całkowite. Od razu widać pewien problem z danymi – w pierwszych linijkach kolumny 'suicides_no', zamiast liczb widać napisy 'null'. Oznacza to, że te wartości są puste.

Specjaliści z zakresu analizy danych napisali już za dużo słów na temat tego, co robić z pustymi wartościami. My je po prostu usuniemy, a wynik takiej operacji zapiszemy do nowej zmiennej.

```python
df_clean = df.drop_nulls()
df_clean.head(6)
```

Tutaj znów musimy zrobić krótką przerwę na uzupełnienie informacji technicznych. Poprzednio użyliśmy napisu `pl.read_csv(path_to_data)`, aby wywować funkcję `read_csv` z biblioteki `pl` (co jest naszym aliasem na `polars`). Teraz użwamy `df.drop_nulls()`, ale `df` nie jest tym razem żadną biblioteką! Jest tabelą typu `DataFrame`, co zatem może znaczyć ta kropka?

Na początek wysarczy wiedziec, że Python pozwala definiować funkcje powiązane z konkretnymi typami obieków, nazywane metodami. Metoda `drop_nulls` jest powiązana z obiektami `DataFrame` i zwraca obiekt `DataFrame` pozbawiony wszystkich wierszy w których występowały wartości `null`. Kropka między `df` a `drop_nulls` wskazuje z której tabeli powinny zostać usunięte wiersze.

Sprawdźmy jakie unikalne wartości kolumny `country` są dostępne w pliku.

```python
unique_countries = df.select("country").unique()
unique_countries
```

Najpierw wybraliśmy z tabeli tylko jedną kolumnę, a następnie pozwoliliśmy wyświetlać tylko unikalne wartości. Ale danych wciąż jest na tyle dużo, że `Polars` domyślnie pomija więcej wierszy. Możemy to obejść iterjąc po nich używając pętli `for`:

```python
for country in unique_countries.rows():
    print(country)
```

Rozwiązanie to wciąż nie wygląda zbyt elegancko, ale nie będziemy się tym teraz przejmować. Był to przede wszystkim pretekst aby pokazać pętlę `for`. Pozwala ona iterować po wieloelementowych obiektach. Jednym z najczęściej spotykanych obiektów tego typu jest lista. Poniżej pokazujemy jak stworzyć bardzo prostą listę elementów i wypisać je używając pętli `for`. Najprostszym sposobem na stworzenie listy jest wypisanie jej elementów oddzielonych przecinkami w nawiasach kwadratowych.

```python
nazwana_lista = ["siekiera", "motyka", "bimber", "szklanka"]
print("Lista potrzebnych przedmiotów:")
for przedmiot in nazwana_lista:
    print(przedmiot)
print("Wypisywanie listy zakończone")
```

Przejdźmy przez każdą linijkę kodu powyżej:
1. Tworzymy listę i zapisujemy ją do zmiennej `nazwana_lista`
1. Wypisujemy `"Lista potrzebnych przedmiotów:"`
1. Inicjujemy pętlę, każdy element listy będzie po kolei przypisywany do zmiennej `przdmiot`.
1. Komendy wynokywane w czasie pętli oddzielone są tabulatorami
1. Po wykonaniu wszystkich przejść przez pętle wykonywany jest kod który znajduje się za nią – wypisanie `Wypisywanie listy zakończone`

Jak widać, kod wykonywany jest w kolejności. Zwróćmy również uwagę na fakt, że funkcja `print` nie wyświelka cudzysłowów w których zamknięte są stringi. Cudzysłowy te są bowiem częścią składni Pythona, a nie samego tekstu.

## Proste operacje na danych

Spróbujmy w końcu wyciągnąć coś użytecznego z danych, zacznijmy od wyfiltrowania wierszych dotyczących polski.

```python
df_poland = df_clean.filter(pl.col("country") == "Poland")
df_poland.head(5)
```

A teraz wiersze dotyczące polskich kobiet.

```python
df_f_poland = df_poland.filter(pl.col("sex") == "female")
df_f_poland
```

Możemy pogrupować dane według roku.

```python
df_f_poland_grouped = df_f_poland.group_by("year").agg(pl.col("suicides_no").sum())
df_f_poland_grouped
```

Argumenty metody `agg` wskazują w jaki sposób mają być grupowane kolumny – w tym wypadku wartości `suicides_no` dla każdego roku są sumowane. 

Zobaczmy jak zmieniała się liczba samobójstw wśród kobiet w kolejnych latach. Będziemy potrzebować do tego biblioteki `plotly`.

```python
import plotly.express as px

fig = px.line(df_f_poland_grouped, x="year", y="suicides_no")
fig.show()
```

Coś jest nie tak, dane rysowane są w kolejności wierszy, musimy je posortować.

```python
df_f_poland_grouped = df_f_poland.group_by("year").agg(pl.col("suicides_no").sum()).sort("year")
df_f_poland_grouped.head(5)
```

```python
fig = px.line(df_f_poland_grouped, x="year", y="suicides_no")
fig.show()
```

Stwórzmy teraz jeden wykres, na którym będzie widać zarówno mężczyzn i kobiety. Jak widać, argumentem `group_by` może być lista

```python
df_poland_grouped = df_poland.group_by(["year", "sex"]).agg(pl.col("suicides_no").sum()).sort("year")
df_poland_grouped.head(5)
```

```python
fig = px.line(
    df_poland_grouped,
    x="year",
    y ="suicides_no",
    color="sex",
    title="Liczba samobójstw w Polsce w latach 1985-2016",
    labels={"year": "Rok", "suicides_no": "Liczba samobójstw", "sex": "Płeć"},
)
fig.show()
```
